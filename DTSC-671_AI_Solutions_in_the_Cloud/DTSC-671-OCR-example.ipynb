```python
# Step 2: Import libraries
import easyocr
from PIL import Image
from IPython.display import display
import os
import pandas as pd
```

```python
# Step 3: Upload a JPG image
from google.colab import files
uploaded = files.upload()

# Get image filename
image_path = list(uploaded.keys())[0]

# See file named: home-inspection-form (handwritten text on form)
# See file named: costco_receipt_2.pdf (printed text on receipt)
# See file named: autobio-form (handwritten text on form)
# See file named: autobio-form-cropped (handwritten text on form)
```

```python
# Step 4: Display the uploaded image
img = Image.open(image_path)
display(img)
```

```python
# Step 5: Initialize EasyOCR reader
reader = easyocr.Reader(['en'])
```

```python
# Step 6: Perform OCR
# This may take a few minutes.

results = reader.readtext(image_path)
```

```python
# Step 7: Extract and save text
extracted_text = "\n".join([res[1] for res in results])

# Step 7: Prepare text output
text_only = "\n".join([res[1] for res in results])
txt_filename = "ocr_output.txt"
with open(txt_filename, "w") as f:
    f.write(text_only)

print(f"Extracted text saved to {txt_filename}.")


```

```python
# Step 8: Prepare CSV output
csv_data = [{
    "bounding_box": res[0],
    "text": res[1],
    "confidence": res[2]
} for res in results]

df = pd.DataFrame(csv_data)
csv_filename = "ocr_output.csv"
df.to_csv(csv_filename, index=False)

# Step 9: Notify + Offer Downloads
print(f"Text saved to: {txt_filename}")
print(f"Detailed OCR results saved to: {csv_filename}")

#files.download(txt_filename)
files.download(csv_filename)
```

```python
# Imports
import pytesseract
from PIL import Image
from google.colab import files
```

```python
# Function to extract text from the uploaded image
def extract_text_from_image(image_path):
    image = Image.open(image_path)
    text = pytesseract.image_to_string(image)
    return text

# Function to save the extracted text to a text file
def save_text_to_txt(text, txt_filename):
    with open(txt_filename, 'w') as txt_file:
        txt_file.write(text)
    print(f"Text saved to {txt_filename}")

# Function to handle the file upload and processing
def upload_and_process_image():
    uploaded = files.upload()
    if uploaded:
        file_path = list(uploaded.keys())[0]
        try:
            extracted_text = extract_text_from_image(file_path)
            save_text_to_txt(extracted_text, 'output.txt')
        except Exception as e:
            print(f"An error occurred: {e}")
    else:
        print("No file uploaded.")

# Upload a file and process it
upload_and_process_image()

print(f"Text saved to: {txt_filename}")
files.download(txt_filename)

```
